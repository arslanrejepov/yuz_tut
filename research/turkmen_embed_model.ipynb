```python
models_to_test = [
    "intfloat/multilingual-e5-small",
    "intfloat/multilingual-e5-base",
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
    "sentence-transformers/LaBSE",
]
```

```python

def evaluate_model(model_name, corpus, queries, use_e5_prefix):
    model = SentenceTransformer(model_name)

    passages = [c["text"] for c in corpus]
    if use_e5_prefix:
        passages = ["passage: " + p for p in passages]
    corpus_emb = model.encode(passages, normalize_embeddings=True)

    hits_at_1 = 0
    hits_at_3 = 0
    mrr_sum = 0.0

    for q in queries:
        qtext = ("query: " + q["query"]) if use_e5_prefix else q["query"]
        q_emb = model.encode([qtext], normalize_embeddings=True)[0]
        sims = corpus_emb @ q_emb
        ranked_idx = np.argsort(-sims)
        ranked_ids = [corpus[i]["id"] for i in ranked_idx]

        expected = set(q["expected_ids"])

        if ranked_ids[0] in expected:
            hits_at_1 += 1
        if expected & set(ranked_ids[:3]):
            hits_at_3 += 1

        for rank, cid in enumerate(ranked_ids, start=1):
            if cid in expected:
                mrr_sum += 1.0 / rank
                break

    n = len(queries)
    return {
        "model": model_name,
        "hit@1": hits_at_1 / n,
        "hit@3": hits_at_3 / n,
        "mrr": mrr_sum / n,
    }


results = []
for m in models_to_test:
    use_prefix = "e5" in m.lower()
    print(f"Evaluating {m} ...")
    res = evaluate_model(m, corpus, queries, use_e5_prefix=use_prefix)
    results.append(res)
    print(res)
    print()

print("=" * 60)
print(f"{'model':55s} hit@1  hit@3  mrr")
for r in results:
    print(f"{r['model']:55s} {r['hit@1']:.2f}   {r['hit@3']:.2f}   {r['mrr']:.2f}")
```
